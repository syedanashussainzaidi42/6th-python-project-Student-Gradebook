```python
import csv 
file_name = "student_grade.csv"
```

```python
def setup():
    with open(file_name, "w", newline = "")as file:
        writer = csv.writer(file)
        writer.writerow(["Name", "Subject", "Marks", "Grade"])

    print("Gradebook file created successfully")    
```

```python
def calculate_grade(marks):
    if marks >= 80:
        return"A"
    elif marks >= 70:
        return"B"
    elif marks >= 60:
        return"C"
    elif marks >= 50:
        return"D"
    else:
        return"F"
```

```python
def add_student():
    name = input("Enter Student Name: ")
    subject = input("Enter Subject: ")
    marks = int(input("Enter Marks: "))
    grade = calculate_grade(marks)

    with open(file_name, "a", newline = "") as file:
        writer = csv.writer(file)
        writer.writerow([name, subject, marks, grade])
    print("Student Record Added Successfully!")    
```

```python
def view_students():
    try:
        with open(file_name, "r", newline = "")as file:
            reader = csv.reader(file)
            next(reader)

            print("\n===== All Student Records =====")
            records_found = False
            for row in reader:
              records_found = True
              print("\nName:",row[0])
              print("Subject:",row[1])
              print("Marks:",row[2])
              print("Garde:",row[3])
            
            if records_found == False:
                print("No Student Records are available.")


    except FileNotFoundError:
        print("Gradebook file was not found.")
        print("Please run the setup cell first.")
            

            
            
```

```python
setup()


while True:
    print("===== STUDENT GRADEBOOK =====")
    print("1. Add student record")
    print("2. View all record")
    print("3. Exit")
    
    choice = input("Enter Your Choice: ").strip()


    if choice == "1":
       add_student()

    elif choice == "2":
         view_students()

    elif choice == "3":
         print("Gradebook Closed.")

    else:
        print("Invalid choice. Please enter 1, 2 or 3")
```
